## Pure Functions
A pure function is a function that does not call other parts of yourcode or other functions within a function. In python all functions should be pure functions

A pure function ensure that you code stays clean and it makes it easy to debuging problems

```python
# true functional programing
def square(alist):
    blist = []
    for items in alist:
        blist.append(items*items)
    return blist
```

```python
#not true fucntional programing
def squares(alist):
    print(square(alist))

a = square([1,2,3,4])
print(a) #a printed py print
b = squares([1,2,3,4]) # b printed by squares
```

A pure function should not call another function
- if something breaks in the code, it causes more problems/confusion for debugging. 
- if all of your code relies on squares and it breaks, you have a lot of issues

# Recursion

```python
from functools import reduce

def accumulate (accum, item):
    print(accum, item)
    return(accum +item)

print(reduce(accumulate,[1,2,3],0))
```

```python
def accumulate(accume, item):
    if accume:
        item =+ accume[0] + accumulate(accume[1:], item) #the function call itself over and over again
    return item                                          #until it reaches the end and calculates everything at once.

accumulate([1,2,3], 0)
```
